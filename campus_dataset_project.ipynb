
#**DATA CLEANING**


Data Cleaning and Type Conversion
I will convert the columns that should contain numeric values like temperature, sound levels, and ratings into actual numeric values. Any non numeric things like units or typos will be cleaned or forced to missing values (NaN).




```python
import pandas as pd
import numpy as np

# Define the columns that we want to convert to numeric
numeric_cols = [
    'temperature_c',
    'sound level',
    'wellbeing_rating',
    'ease_rating',
    'stress_rating',
    'nature_connection_experience_rating'
]
```

##location cleaning

```python
# 8. Clean and standardize the 'location' column
df_clean = df.copy()

# Standardize text by stripping whitespaces and converting to title case
df_clean['location'] = df_clean['location'].astype(str).str.strip().str.title()

# Replace any standard representations of NaN string back to actual NaN
df_clean['location'] = df_clean['location'].replace({'Nan': np.nan, 'None': np.nan})

# Display unique categories in the cleaned column
print("Cleaned Location Categories:")
print(df_clean['location'].value_counts(dropna=False))

# Compare a preview
display(pd.DataFrame({
    'Original Location': df['location'],
    'Cleaned Location': df_clean['location']
}).head(10))
```

## sound level

```python
# 6. Clean and standardize the 'sound level' column
df_clean = df.copy()

# Standardize text by stripping whitespaces and converting to title case
df_clean['sound level'] = df_clean['sound level'].astype(str).str.strip().str.title()

# Replace any standard representations of NaN string back to actual NaN
df_clean['sound level'] = df_clean['sound level'].replace({'Nan': np.nan, 'None': np.nan})

# Display unique categories in the cleaned column
print("Cleaned Sound Level Categories:")
print(df_clean['sound level'].value_counts(dropna=False))

# Compare a preview
display(pd.DataFrame({
    'Original Sound Level': df['sound level'],
    'Cleaned Sound Level': df_clean['sound level']
}).head(10))
```

##**temperature cleaning**

```python
# 1. Clean and convert 'temperature_c'
df_clean = df.copy()

# Convert to string, strip whitespace, remove any non-numeric characters (like 'C' or '°'), and convert to float
df_clean['temperature_c'] = pd.to_numeric(
    df_clean['temperature_c'].astype(str).str.replace(r'[^0-9.-]', '', regex=True),
    errors='coerce'
)

# Display the cleaned values alongside the original ones to verify
display(pd.DataFrame({
    'Original Temperature': df['temperature_c'],
    'Cleaned Temperature': df_clean['temperature_c']
}).head(10))
```

##**date cleaning**

```python
# 3. Clean and convert 'date'
# Convert to standard datetime format, letting pandas infer the day/month/year order automatically
df_clean['date'] = pd.to_datetime(df_clean['date'], errors='coerce')

# Display the cleaned dates alongside the original ones to verify
display(pd.DataFrame({
    'Original Date': df['date'],
    'Cleaned Date': df_clean['date']
}).head(10))
```

##**weather cleaning**

```python
# 4. Clean and standardize the 'weather' column
# Standardize text by stripping whitespaces and converting to title case (e.g., 'Sunny', 'Partly Cloudy')
df_clean['weather'] = df_clean['weather'].astype(str).str.strip().str.title()

# Replace any standard representations of NaN string back to actual NaN
df_clean['weather'] = df_clean['weather'].replace({'Nan': np.nan, 'None': np.nan})

# Display unique categories in the cleaned column
print("Cleaned Weather Categories:")
print(df_clean['weather'].value_counts(dropna=False))

# Compare a preview
display(pd.DataFrame({
    'Original Weather': df['weather'],
    'Cleaned Weather': df_clean['weather']
}).head(10))
```

##**cleaned other notes as there is only a few**

```python
# 5. Clean and standardize the 'other_notes' column
# Fill NaN values with an empty string temporarily to allow text operations, or keep them as NaN
df_clean['other_notes'] = df_clean['other_notes'].astype(str).str.strip()

# Convert standard placeholder strings back to true NaN
placeholder_strings = ['nan', 'none', 'n/a', '', 'no notes', 'nil']
df_clean['other_notes'] = df_clean['other_notes'].apply(lambda x: np.nan if str(x).lower() in placeholder_strings else x)

# Capitalize first letter of each note for clean presentation
df_clean['other_notes'] = df_clean['other_notes'].apply(lambda x: str(x)[0].upper() + str(x)[1:] if pd.notnull(x) else x)

# Compare a preview of original and cleaned notes
display(pd.DataFrame({
    'Original Notes': df['other_notes'],
    'Cleaned Notes': df_clean['other_notes']
}).dropna(subset=['Cleaned Notes']).head(10))
```

# Encoding
using One Hot Encoding (pd.get_dummies) to convert the categorical `weather` column into numerical format.

## sound level encoding

```python
# 7. Encode the 'sound level' column using integer/label encoding

sound_mapping = {
    'Low': 1,
    'Medium': 2,
    'High': 3
}

# Map sound level categories to numbers
df_clean['sound_level_encoded'] = df_clean['sound level'].map(sound_mapping)

# Display encoded sound level values
print("Encoded Sound Level Categories:")
print(df_clean['sound_level_encoded'].value_counts(dropna=False))

# Compare original and encoded values
display(df_clean[['sound level', 'sound_level_encoded']].head(15))
```

##Weather encoding

```python
# 5. Encode the 'weather' column using integer/label encoding

if 'weather' in df_clean.columns:

    # Define weather mapping
    weather_mapping = {
        'Sunny': 1,
        'Partly Cloudy': 2,
        'Cloudy': 3,
        'Windy': 4
    }

    # Map weather categories to numbers
    df_clean['weather_encoded'] = df_clean['weather'].map(weather_mapping)

    # Display encoded weather values
    print("Encoded Weather Categories:")
    print(df_clean['weather_encoded'].value_counts(dropna=False))

    # Compare original and encoded values
    display(df_clean[['weather', 'weather_encoded']].head(15))

else:
    print('Weather column not found')
```

##location encoding

```python
# 9. Encode the 'location' column using integer/label encoding

# Define location mapping
location_mapping = {
    'L01 Mainentrance': 1,
    'L02 Librarygreen': 2,
    'L03 Lowersidepathihub': 3,
    'L04 Pathfrontihub': 4,
    'L05 Greenfrontsportscentre': 5,
    'L06 Hedgingbehindsportscentre': 6,
    'L07 Carparknearsportscentre': 7,
    'L08 Accessroadnearsportscentre': 8
}

# Map location categories to numbers
df_clean['location_encoded'] = df_clean['location'].map(location_mapping)

# Display encoded location values
print("Encoded Location Categories:")
print(df_clean['location_encoded'].value_counts(dropna=False))

# Compare original and encoded values
display(df_clean[['location', 'location_encoded']].head(15))
```
